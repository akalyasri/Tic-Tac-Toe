# Advanced Tic-Tac-Toe AI Demo Notebook
This notebook simulates Tic-Tac-Toe games between an AI (Minimax with Alpha-Beta Pruning) and a random player, across multiple board sizes.

---

```python

import ttt
import random
import matplotlib.pyplot as plt
    
```

### Simulate a Single Game

```python

def simulate_game(board_size=3, depth_limit=5):
    board = ttt.initial_state(board_size)
    k = board_size
    moves_made = 0

    while not ttt.terminal(board, k):
        current_player = ttt.player(board)
        if current_player == ttt.X:
            move = ttt.minimax(board, k, depth_limit)
        else:
            possible_moves = list(ttt.actions(board))
            move = random.choice(possible_moves)
        board = ttt.result(board, move)
        moves_made += 1

    winner = ttt.check_winner(board, k)
    return winner, moves_made
    
```

### Simulate Multiple Games and Collect Statistics

```python

def run_simulation(num_games=10, board_size=3, depth_limit=5):
    results = {"X": 0, "O": 0, "Draw": 0}
    total_moves = 0

    for _ in range(num_games):
        winner, moves = simulate_game(board_size, depth_limit)
        if winner == ttt.X:
            results["X"] += 1
        elif winner == ttt.O:
            results["O"] += 1
        else:
            results["Draw"] += 1
        total_moves += moves

    avg_moves = total_moves / num_games
    return results, avg_moves
    
```

### Run Simulations Across Multiple Board Sizes

```python

# Board sizes and depth settings
test_cases = [
    (3, 4),  # (board_size, depth_limit)
    (4, 4),
    (5, 5),
    (8, 3),
    (10, 3),
    (12, 2)  # shallower depth due to complexity
]

all_results = {}

for board_size, depth_limit in test_cases:
    results, avg_moves = run_simulation(num_games=20, board_size=board_size, depth_limit=depth_limit)
    win_rate = results["X"] / 20
    all_results[board_size] = {
        "win_rate": win_rate,
        "avg_moves": avg_moves,
        "results": results
    }

# Display results
for size in sorted(all_results.keys()):
    print(f"\nBoard Size: {size}x{size}")
    print(f"Win Rate (AI X): {all_results[size]['win_rate']*100:.1f}%")
    print(f"Average Moves: {all_results[size]['avg_moves']:.1f}")
    print(f"Detailed Results: {all_results[size]['results']}")
    
```

### Visualize Results

```python

# Plotting Win Rate vs Board Size
sizes = sorted(all_results.keys())
win_rates = [all_results[size]["win_rate"] for size in sizes]

plt.figure(figsize=(8,5))
plt.plot(sizes, win_rates, marker='o')
plt.title('AI Win Rate vs Board Size')
plt.xlabel('Board Size (NxN)')
plt.ylabel('Win Rate (0 to 1)')
plt.grid(True)
plt.show()

# Plotting Average Moves vs Board Size
avg_moves_list = [all_results[size]["avg_moves"] for size in sizes]

plt.figure(figsize=(8,5))
plt.plot(sizes, avg_moves_list, marker='o', color='green')
plt.title('Average Moves per Game vs Board Size')
plt.xlabel('Board Size (NxN)')
plt.ylabel('Average Moves')
plt.grid(True)
plt.show()
    
```
